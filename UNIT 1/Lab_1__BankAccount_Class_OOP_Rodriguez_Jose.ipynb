```python
class window:
    def __init__(self, color, material, size):
        self.color = color
        self.material = material
        self.size = size

    def open(self):
            print("The window is open.")

    def close(self):
            print("The window is closed.")
    def describe(self):
            print(f"This window is {self.color}, made of {self.material}, and is {self.size} in size.")
# Create multiple instances of the window class
window1 = window("white", "wood", "2x3 feet")
window2 = window("black", "aluminum", "3x4 feet")
print(window1.material)  # Output: wood
print(window2.material)  # Output: aluminum
window1.describe()  # Output: This window is white, made of wood, and is 2x3 feet in size.
window2.describe()  # Output: This window is black, made of aluminum, and is 3x4 feet in size.
```

```python
class BankAccount:
    def __init__(self, holder, initial_balance):
        self.holder = holder
        self.__balance = initial_balance
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposit successful. Added ${amount}.")
        else:
            print("Error: Deposit amount must be greater than zero.")
    def withdraw(self, amount):
        if amount > self.__balance:
            print("Error: Insufficient funds. Cannot withdraw more than the available balance.")
        elif amount <= 0:
            print("Error: Withdrawal amount must be greater than zero.")
        else:
            self.__balance -= amount
            print(f"Withdrawal successful. Removed ${amount}.")
    def check_balance(self):
        print(f"Current balance: ${self.__balance}")
account1 = BankAccount("Raúl Pérez", 5000)
account2 = BankAccount("Joel López", 3000)
def test_account(account, deposit_amount, withdraw_amount):
    print(f"--- Account Info ---")
    print(f"Account Holder: {account.holder}")
    account.check_balance()
    print(f"\nAttempting to deposit ${deposit_amount}...")
    account.deposit(deposit_amount)
    print(f"\nAttempting to withdraw ${withdraw_amount}...")
    account.withdraw(withdraw_amount)
    print("\nFinal Status:")
    account.check_balance()
    print("\n")
test_account(account1, deposit_amount=1500, withdraw_amount=2000)
test_account(account2, deposit_amount=500, withdraw_amount=4000)
```
